```python
#How should we replace the null value?
```

```python
#Through Mean or Median
#Generally medain > Mean
#Because meadian doesnt get affected by outliars 
#common mistake when we do is like find the median of entire column and replace it with null value , which is technacilly correct but not the ideal one.
#Ratings can vary a lot
#diff categories can have different rating which can force bias into our dataset
```

```python
#We Will impute median into review rating within each categories
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#we can see that the column names are in different cases and have spaces between them , which can cause multiple problems when analysing the dataset
```

```python
#Snake_cases
#eg: customer_id
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create a column age_group


labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
purchases = df['frequency_of_purchases'].tolist()
print(purchases)
```

```python
#create purchase_frequency_days

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
